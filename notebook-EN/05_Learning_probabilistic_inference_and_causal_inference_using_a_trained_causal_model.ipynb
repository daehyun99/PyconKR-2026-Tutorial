````python

import pandas as pd
import numpy as np

from pgmpy_tutorial.helper_functions import load_college_plans

data = load_college_plans()
````

````python
# Parameter Learning
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.parameter_estimator import DiscreteMLE

model = DiscreteBayesianNetwork()
model.add_edge("sex", "pe", "->")
model.add_edge("ses", "pe", "->")
model.add_edge("ses", "iq", "->")
model.add_edge("iq", "pe", "->")
model.add_edge("ses", "cp", "->")
model.add_edge("iq", "cp", "->")
model.add_edge("pe", "cp", "->")

model = model.fit(data, estimator=DiscreteMLE())
````

## 5.4. Inference

```py
sex = { # Sex
    "1": male,
    "2": female,
}
iq = { # Intelligence
    "1": "Low",
    "2": "Lower middle",
    "3": "Upper middle",
    "4": "High",
}
cp = { # College plans
    "1": "Yes",
    "2": "No,
}
pe = { # Parental encouragement (modified from the original data)
    "1": "High",
    "2": "Low",
}
iq = { 
    "1": "Low",
    "2": "Lower middle",
    "3": "Upper middle",
    "4": "High",
}
```

![05](../image/05-004.excalidraw.png)

````python
from pgmpy.inference import CausalInference
inference = CausalInference(model)
result = inference.query(variables=["ses"], evidence={"pe":2, "iq":1, "cp":1, "sex":1})
print(result)
````
