#### 1.2.2. Does a strong correlation make prediction possible?


```python
# Import
import pandas as pd
import numpy as np

# Setting
N_SAMPLES = 1000
```

```python
from pgmpy_tutorial.helper_functions import load_stevie_netflix_dataset

array_1, array_2 = load_stevie_netflix_dataset()
```

```python
array_1 = (array_1 - array_1.min()) / (array_1.max() - array_1.min())
array_2 = (array_2 - array_2.min()) / (array_2.max() - array_2.min())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(array_1, array_2, train_size=0.8, shuffle=True, random_state=42)
```

```python
X_train = X_train.reshape(-1, 1)
X_test = X_test.reshape(-1, 1)
y_train = y_train.reshape(-1, 1)
y_test = y_test.reshape(-1, 1)
```

```python
from sklearn.linear_model import LinearRegression

estimator = LinearRegression()
estimator.fit(X_train, y_train)
```

```python
y_pred = estimator.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5

print("MSE:", mse)
print("RMSE:", rmse)
```

```python
rmse = 25.136605427601342
mean_y = y_test.mean()

nrmse = rmse / mean_y * 100
print(nrmse)
```

> Why this works<br>
> 1. Data dredging: I have 25,153 variables in my database. I compare all these variables against each other to find ones that randomly match up. That's 632,673,409 correlation calculations! This is called “data dredging.” Instead of starting with a hypothesis and testing it, I instead abused the data to see what correlations shake out. It’s a dangerous way to go about analysis, because any sufficiently large dataset will yield strong correlations completely at random.<br>
> 2. Lack of causal connection: There is probablyNote no direct connection between these variables, despite what the AI says above. This is exacerbated by the fact that I used "Years" as the base variable. Lots of things happen in a year that are not related to each other! Most studies would use something like "one person" in stead of "one year" to be the "thing" studied. <br>
> 3. Observations not independent: For many variables, sequential years are not independent of each other. If a population of people is continuously doing something every day, there is no reason to think they would suddenly change how they are doing that thing on January 1. A simpleNote p-value calculation does not take this into account, so mathematically it appears less probable than it really is.<br>
