#### 2.2.2. Common Graph Features Supported by `pgmpy`

> Graph methods are designed to begin with prefixes such as `add_*`, `get_*`, `has_*`, `is_*`, and `to_*`.

- `add_edge(u, v, edge_type)`: Adds an edge to the graph.
    - `u`, `v`: Nodes
    - `edge_type`: Type of edge
        - "->", "<-": Directed edge
        - "--": Edge with no direction at either end
        - "<>": Edge directed at both ends
        - "-o", "o-": Edge with no direction at one end and an unknown direction at the other
        - "o>", "<o": Edge with a direction at one end and an unknown direction at the other
        - "oo": Edge with unknown directions at both ends

- `remove_edge(u, v, edge_type)`: Removes an edge from the graph.
    - `u`, `v`: Nodes
    - `edge_type`: Type of edge
        - "->", "<-": Directed edge
        - "--": Edge with no direction at either end
        - "<>": Edge directed at both ends
        - "-o", "o-": Edge with no direction at one end and an unknown direction at the other
        - "o>", "<o": Edge with a direction at one end and an unknown direction at the other
        - "oo": Edge with unknown directions at both ends

- `get_edges(data, edge_type)`: Retrieves information about the graph's edges.
    - `data`: Whether to include edge-direction information in the output
    - `edge_type`: Retrieve only a specific type of edge

- `get_roots()`, `get_leaves()`: Retrieve the top-level and bottom-level nodes of the graph.

- `get_neighbors(node, edge_types)`: Retrieves neighboring nodes of `node`.

- `get_parents(nodes)`, `get_children(nodes)`, `get_spouses(nodes)`, `get_ancestors(nodes)`, `get_descendants(nodes)`: Retrieve nodes connected to `node` by specific edge types.

- `get_reachable_nodes(node, edge_types)`: Retrieves nodes reachable from `node` along paths composed of specific edge types.

- `is_collider(u, w, v)`: Checks whether the path through three adjacent nodes `u`, `w`, and `v` forms a collider.

- `do()`: Returns the graph produced by applying the `do` operation to the existing graph.

- `is_mseparated(u, v, conditioning_set)`: Checks whether two nodes `u` and `v` are conditionally independent given `conditioning_set`.

- `has_path(source, target, edge_types)`: Checks whether a path composed of `edge_types` exists from the `source` node to the `target` node.

### 2.4. M-Separation, the Markov Assumption, and the Faithfulness Assumption

![02](../image/02-003.excalidraw.png)


```python
ChainGraph = Graph()

ChainGraph.add_edge("A", "B", "->")
ChainGraph.add_edge("B", "C", "->")


pgm = ChainGraph.to_daft(node_pos="circular")

pgm.render()
```

```python
rng = np.random.default_rng(seed=42)
A = rng.random(size=N_SAMPLES)
B = A + rng.random(size=N_SAMPLES)
C = B + rng.random(size=N_SAMPLES)

data = pd.DataFrame(
    {
        "A": A,
        "B": B,
        "C": C,
    }
)
```

```python
from pgmpy.ci_tests import Pearsonr

test = Pearsonr(data=data)

assert test(X="A", Y="C", Z=[], significance_level=0.05) is ChainGraph.is_mseparated("A", "C")

assert test(X="A", Y="C", Z=["B"], significance_level=0.05) is ChainGraph.is_mseparated("A", "C", "B")

```

```python
ForkGraph = Graph()

ForkGraph.add_edge("A", "B", "<-")
ForkGraph.add_edge("B", "C", "->")

pgm = ForkGraph.to_daft(node_pos="circular")

pgm.render()

```

```python
rng = np.random.default_rng(seed=42)
B = rng.random(size=N_SAMPLES)
A = B + rng.random(size=N_SAMPLES)
C = B + rng.random(size=N_SAMPLES)

data = pd.DataFrame(
    {
        "A": A,
        "B": B,
        "C": C,
    }
)
```

```python
from pgmpy.ci_tests import Pearsonr

test = Pearsonr(data=data)

assert test(X="A", Y="C", Z=[], significance_level=0.05) is ForkGraph.is_mseparated("A", "C")

assert test(X="A", Y="C", Z=["B"], significance_level=0.05) is ForkGraph.is_mseparated("A", "C", "B")
```

```python
ColliderGraph = Graph()

ColliderGraph.add_edge("A", "B", "->")
ColliderGraph.add_edge("B", "C", "<-")

pgm = ColliderGraph.to_daft(node_pos="circular")

pgm.render()

```

```python
rng = np.random.default_rng(seed=42)
A = rng.random(size=N_SAMPLES)
C = rng.random(size=N_SAMPLES)
B = A + C + rng.random(size=N_SAMPLES)

data = pd.DataFrame(
    {
        "A": A,
        "B": B,
        "C": C,
    }
)
```

```python
from pgmpy.ci_tests import Pearsonr

test = Pearsonr(data=data)

test(X="A", Y="C", Z=[], significance_level=0.05) is ColliderGraph.is_mseparated("A", "C")

test(X="A", Y="C", Z=["B"], significance_level=0.05) is ColliderGraph.is_mseparated("A", "C", "B")
```
